```python
# Read a PDF file using python
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
print("Number of Pages:",len(reader.pages))
#Read first page
page = reader.pages[0]
text = page.extract_text()
print(text)
```

```python
#Read all pages from a PDF
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
for page_number, page in enumerate(reader.pages,start = 1):
    print(f"/n--- Page{page_number}---")
    text = page.extract_text()
    print(text)
```

```python
# Read a Specific page
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
page = reader.pages[2]
print(page.extract_text())
```

```python
#Search for a word in a PDF
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
search_word = "hence"
for page_number, page in enumerate (reader.pages, start = 1):
    text = page.extract_text()
    if search_word.lower() in text.lower():
        print(f"'{search_word}' found on page number {page_number}")
```

```python
#Count Words in a PDf
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
word_count = 0
for page in reader.pages:
    text = page.extract_text()
    if text:
        word_count += len(text.split())
print("Total words:", word_count)
```

```python
#copy the first page into another PDF
from PyPDF2 import PdfReader, PdfWriter
#Read original PDf
reader = PdfReader("book.pdf")
#Create new Pdf
writer = PdfWriter()
#Copy first page
writer.add_page(reader.pages[0])
#Save as another PDf
with open ("first_page.pdf","wb") as output:
    writer.write(output)
print("copied successfully!")
```

```python
#save extracted text into a .txt file
from PyPDF2 import PdfReader
reader = PdfReader("book.pdf")
with open("pdf_text.txt","w") as file:
    for page_number, page in enumerate (reader.pages,start = 1):

        text = page.extract_text()
        file.write(f"/n--- Page {page_number} ---/n")
        file.write(text if text else"")
print("PDF text saved successfully")
```
