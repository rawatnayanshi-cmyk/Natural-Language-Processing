```python
#3. Extract only stopwords.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "This is a how we remove stopword."
doc = nlp(text)
stopwords = []
for token in doc:
  if token.is_stop:
    stopwords.append(token.text)
print(stopwords)
```

```python
#4. Extract Non- Stopwords.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy programming language."
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print(words)
```

```python
#5. Remove stopwords from a sentence.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "These students are studying python in class."
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop:
    result.append(token.text)
print(" ".join(result))
```

```python
#6. Remove stopwords and Punctuation.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "There are so many people in the concert,The concert was great."
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text)
print(" ".join(result))
```

```python
#7. Convert text to lowercase and remove stopwords.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Capital of India is Delhi."
doc = nlp(text)
result = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text.lower())
print(" ".join(result))
```

```python
#8. Display spacy's stopword list.
import spacy
nlp = spacy.load("en_core_web_sm")
stopwords = nlp.Defaults.stop_words
print(stopwords)
len(stopwords)
```

```python
#9. Add a Custom Stopword.
import spacy
nlp = spacy.load("en_core_web_sm")
nlp.Defaults.stop_words.add("read")
doc = nlp("There are so many books to read.")
for token in doc:
  print(token.text, "->", token.is_stop)

```

```python
#10. Create a stopword removal function.
import spacy
nlp = spacy.load("en_core_web_sm")
def remove_stopwords(text):
  doc = nlp(text)
  result = []
  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.text)
  return " ".join(result)
text = "Focus is important for learning something new."
print(remove_stopwords(text))
```

```python
#3. Extract stopwords from a sentence.
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download("punkt_tab")
stop_words = set(stopwords.words('english'))
text = "The food is yummy."
tokens = word_tokenize(text)
result = []
for token in tokens:
  if token in stop_words:
    result.append(token)
print(result)
```

```python
#4. Remove stopwords from a sentence.
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download("punkt_tab")
stop_words = set(stopwords.words('english'))
text = "Rishabh is a intelligent boy."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(" ".join(result))
```

```python
#5. Remove stopwords and Punctuations.
import string
stop_words = set(stopwords.words('english'))
text = "Water is essential for live"
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)
print(" ".join(result))
```

```python
#6. Convert text to lowercase and remove stopwords.
stop_words = set(stopwords.words('english'))
text = "Show an example of training a simple ML model."
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print(" ".join(result))
```

```python
#7. Count stopwords in a text.
text = "How can I install Python libraries?"
tokens = word_tokenize(text)
stop_words = set(stopwords.words('english'))
count = 0
for word in tokens:
  if word.lower() in stop_words:
    count += 1
print(count)
```

```python
#9. Add custom Stopwords in NLTK.
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
stop_words = set(stopwords.words('english'))
custom_words = {"python","data","language"}
stop_words.update(custom_words)
text = "Load data from google drive."
from nltk.tokenize import word_tokenize
nltk.download("punkt")
nltk.download("punkt_tab")
tokens = word_tokenize(text)
result = []
for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print(" ".join(result))
```
