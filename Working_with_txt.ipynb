```python
#6. Extract only stopwords.
import nltk
from nltk.corpus import stopwords
# Download required resources
nltk.download("punkt")
nltk.download("stopwords")
# Step 1: Read the text file
with open("story.txt", "r", encoding="utf-8") as f:
    story = f.read()
# Step 2: Tokenize into words
tokens = nltk.word_tokenize(story)
# Step 3: Get English stopwords
stop_words = set(stopwords.words("english"))
# Step 4: Extract only stopwords from the text
story_stopwords = [w for w in tokens if w.lower() in stop_words]
print("Stopwords found in story.txt:")
print(story_stopwords[:100])  # preview first 50 stopwords
```

```python
#7. Import non- Stopwords
import nltk
from nltk.corpus import stopwords

# Download resources if not already
nltk.download("punkt")
nltk.download("stopwords")
# Step 1: Read the text file
with open("story.txt", "r", encoding="utf-8") as f:
    story = f.read()
# Step 2: Tokenize into words
tokens = nltk.word_tokenize(story)
# Step 3: Get English stopwords
stop_words = set(stopwords.words("english"))
# Step 4: Extract only non-stopwords (meaningful words)
non_stopwords = [w for w in tokens if w.lower() not in stop_words and w.isalpha()]
print("Non-stopwords found in story.txt:")
print(non_stopwords[:50])  # preview first 50 non-stopwords
```

```python
#8. Removing stopwords
import nltk
from nltk.corpus import stopwords

# Download resources if not already
nltk.download("punkt")
nltk.download("stopwords")

# Step 1: Read the text file
with open("story.txt", "r", encoding="utf-8") as f:
    story = f.read()

# Step 2: Tokenize into words
tokens = nltk.word_tokenize(story)

# Step 3: Get English stopwords
stop_words = set(stopwords.words("english"))

# Step 4: Remove stopwords
filtered_tokens = [w for w in tokens if w.lower() not in stop_words and w.isalpha()]

print("Text without stopwords:")
print(filtered_tokens[:50])  # preview first 50 words
```
