```python
#Stemming in NLP
# Stemming is the process of reducing words to their root/base form (Stem) by removing prefixes and suffix.
```

```python
# Import porterStemmer
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
#1. Basic Stemming
words = ["singing","singed","sings","singer"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#2. Stemming  a sentence
sentence = "The children are playing in the park."
words = sentence.split()
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#3. Words ending with "ing".
words = ["growing","learning","singing","sunging"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#4. Words ending with "ed".
words = ["locked","rocked","talked","completed"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#5. Words ending with "s".
words = ["cuteness","dogs","books","students"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#6. Different forms of the same word
words = ["want","wanted","wanting","situation","situations"]
for word in words:
    print(word,"->",stemmer.stem(word))
```

```python
#7. Stemming with the loop
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ["loved","loves","loving"]
for word in words:
  stem = stemmer.stem(word)
  print(f"{word:12}->{stem}")
```

```python
#8. Stemming words related to orgaize
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ["organize","organizes","organized","organization"]
for word in words:
  stem = stemmer.stem(word)
  print(f"{word:12}->{stem}")
```

```python
#9. Stemming text using regular expressions.
import re
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
text = "(NLP) is a branch of AI that enables computers to understand generate human language for tasks like translation."
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
for word in words:
  print(word, "->", stemmer.stem(word))
```

```python
# import the libraries.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
```

```python
#1. Stemming spacy tokens.
text = "The children are playing."
doc = nlp(text)
for token in doc:
  if token.is_alpha:

    print(token.text, "->", stemmer.stem(token.text))
```

```python
#2. cooking,cooked,cooks.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
text = "I cook food.He cooks food.They cooked yesterday."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
#4. playing-played words.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
text = "The athletes are playing played matches."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(token.text, "->", stemmer.stem(token.text))
```

```python
#5. Stemming a complete paragraph.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
text = "There is darkness in night."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

```python
#6. Remove Punctuation before stemming.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
text = "Python is an important language for a coder, to learn."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

```python
#7. Stemming with lowercase conversion.
text = "Cook Cooks Cooking."
doc = nlp(text.lower())
for token in doc:
  if token.is_alpha:
    word = token.text
    stemmed_word = stemmer.stem(word)
    print(f"{word:15} -> {stemmed_word}")
```

```python
#8. Create a stemmed sentence using SpaCy.
import spacy
from nltk.stem import PorterStemmer
nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
text = "The children are playing games and enjoying themselves."

doc = nlp(text)
stemmed_words = []
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))
stemmed_sentence = ' '.join(stemmed_words)
print("Original Sentence:", text)
print("\nStemmed Sentence:", stemmed_sentence)
```
