```python
# Creating and Writing a text file.
file = open("NLP.txt","w")
file.write("Natural Language Processing (NLP) is a field of Artificial Intelligence that focuses on enabling computers to understand, interpret, and generate human language by combining linguistics, computer science, and machine learning; it involves tasks such as text preprocessing, tokenization, sentiment analysis, translation, and speech recognition, and is widely applied in chatbots, search engines, voice assistants, spam filters, and recommendation systems using techniques that range from rule‑based methods and statistical models to advanced deep learning approaches like BERT and GPT.")

file.close()
print("File Created Successfully.")
```

```python
# writing multiple lines using writelines()
Topics =["Lemmatization\n",
          "Stemming\n"
]
with open("NLP.txt","w") as file:
    file.writelines(Topics)
print("Data written successfully.")
```

```python
# Reading the entire file
with open("NLP.txt","r") as file:
    data = file.read()
print(data)
```

```python
# Reading the file line by line
with open("NLP.txt","r") as file:
    for line in file:
        print(line.strip())
```

```python
# Using readlines()
with open("NLP.txt","r") as file:
    lines= file.readlines()

print(lines)
```

```python
# You can also process the list
with open ("NLP.txt","r") as file:
     lines= file.readlines()
for line in lines:
         print(line.strip())
```

```python
# Appending data to a file
with open("NLP.txt","a" )as file:
    file.write("Tokenization\n")
    file.write("Stopwords\n")
print("New Topics added.")
```

```python
# Counting number of lines
with open("NLP.txt","r") as file:
    lines = file.readlines()
print("Number of lines:", len(lines))
```

```python
# Counting words in text file
with open("NLP.txt","r") as file:
    data = file.read()
words = data.split()
print("Number of words:", len(words))
```

```python
# Counting Characters
with open("NLP.txt","r") as file:
    data = file.read()
print("Number of Characters:",len(data))
```

```python
# Searching for a word
search_word = input("Enter word to search:")
with open("NLP.txt","r") as file:
    data = file.read()
if search_word.lower() in data.lower():
    print("Word found.")
else:
    print("Word not found")
```

```python
# counting occurences of a word
search_word = input("Enter Word:")
with open("NLP.txt","r") as file:
    data = file.read()
count = data.lower().count(search_word.lower())
print("Occurrences:",count)
```

```python
# Copy contents from one file to another
with open("NLP.txt","r") as source:
    data = source.read()
with open("NewFile.txt","w") as destination:
    destination.write(data)
print("File Copied Sucessfully")
```

```python
# reading only the first 10 characters
with open("NLP.txt","r") as file:
    data = file.read(10)
print(data)
```
