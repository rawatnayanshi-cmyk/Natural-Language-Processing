```python
#2. Import spacy download the english model.
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
# .Basic Lemmatization
text = "The students are studying in class."
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

```python
# ♦2. Lemmatization Plural Nouns.
text = "Books birds gifts boys."
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

```python
#3. Lemmatizing verbs.
text = "go went gone"
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

```python
#4. Lemmatizing different verb forms.
text = "I am playing.She played chess.He will play basketball."
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

```python
#5.Lemmetization with sentence.

text = "The teachers are teaching their lesson."
doc = nlp(text)
lemmatized_text = " ".join([token.lemma_ for token in doc])
print(lemmatized_text)
```

```python
#6. Lemmatization of Irregular Words.
text = "Go now going going gone"
doc = nlp(text)
for token in doc:
  print(token.text,"->",token.lemma_)
```

```python
#7. Lemmatization based on Part of speech.
text = "The children are playing nicely"
doc = nlp(text)
for token in doc:
  print(token.text, "POS:",token.pos_,
        "Lemma:",token.lemma_)
```

```python
#8. Remove punctuation and lemmatize.
text = "Boys are dancing girls are singing!"
doc = nlp(text)
words = []
for token in doc:
  if not token.is_punct:
    words.append(token.lemma_)
print(words)
```

```python
#9. Lemmatization with Stopword Removal.
text = "The students are learning Python."
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)
print(words)
```

```python
#10. Create a Lemmatization function.
def lemmatize_text(text):
  doc = nlp(text)
  result = []
  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.lemma_)
  return " ".join(result)
text = "The dogs are barking."
result = lemmatize_text(text)
print(result)
```

```python
import nltk
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
#1. Basic NLTK Lemmatization.
words = ["buses","books","colors","houses"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word))
```

```python
#2. Lemmatizing verbs
words = ["go","went","gone"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
#3♫. Lemmatizing Nouns.
words = ["women","mouse","girrafe","men","girl"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="n"))

```

```python
#4. Lemmatizing Adjectives
words = ["better","best","faster","happier"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="a"))
```

```python
#5.Lemmatizing Adverbs.
words = ["normally","hopefully","happily"]
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="r"))
```

```python
#6♫. Irregular Verb Lemmatization.
words = ["went","gone","going","eaten","ate"]
for word in words:
  print(word,"->",lemmatizer.lemmatize(word,pos="v"))
```

```python
#7. Compare Noun and Verb Lemmatization.
#The Same word can produce different results depending on its part of speech.

words = "meeting"
for word in words:
  print(word, "->", lemmatizer.lemmatize(word, pos="n"))
  print(word, "->", lemmatizer.lemmatize(word, pos="v"))
```

```python
#8. Lemmatizing a list of words.
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
words = ["studies","studying","studied","students","books"]
for word in words:
  if word.endswith("ing") or word.endswith("ed"):
    lemma = lemmatizer.lemmatize(word, pos="v")
  else:
    lemma = lemmatizer.lemmatize(word, pos="n")
  print(word, "->", lemma)
```

```python
#9. Lemmatization with Tokenization
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
nltk.download("punkt")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("punkt_tab")
lemmatizer = WordNetLemmatizer()
text = "The teachers are teaching their lessons."
tokens = word_tokenize(text)
for token in tokens:

    lemma = lemmatizer.lemmatize(token)
    print(token, "->", lemma)
```

```python
#10 Complete NLTK Lemmatization with POS
import nltk
from nltk import word_tokenize, pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
nltk.download("punkt")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger")
nltk.download('averaged_perceptron_tagger_eng')
lemmatizer = WordNetLemmatizer()
def get_wordnet_pos(word):
    tag = nltk.pos_tag([word])[0][1]
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN
text = "The girls are playing in the garden."
tokens = word_tokenize(text)
for token in tokens:
    if word.isalpha():
      pos = get_wordnet_pos(word)
      lemma = lemmatizer.lemmatize(token, pos=pos)
      print(token, "->", lemma)
```
